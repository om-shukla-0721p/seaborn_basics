```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
tips_data=sns.load_dataset("tips")
titanic=sns.load_dataset("titanic")
titanic.head()
```

```python
titanic.sex=titanic.sex.apply(lambda x:1 if x=="male" else 0)
titanic.alive=titanic.alive.apply(lambda x:1 if x=="yes" else 0)
titanic.alone=titanic.alone.apply(lambda x:1 if x=="True" else 0)

titanic["class"]=titanic["class"].astype\
    (pd.CategoricalDtype(categories=["First","Second","Third"],ordered=True)).cat.codes
titanic=titanic.drop(["embarked","who","adult_male","deck","embark_town"],axis=1)
titanic.head()
```

```python
titanic.corr()
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(titanic.corr(),annot=True,cmap="coolwarm")
```
