```python
import seaborn as sns
tips_data=sns.load_dataset("tips")
dow_data=sns.load_dataset("dowjones")
tips_data.head()
```

```python
sns.displot(tips_data,col="sex",x="tip",kind="hist")
```

```python
sns.histplot(tips_data,x="tip",hue="sex",kde=True)
sns.rugplot(tips_data,x="tip") # show actual data place in the bin
```

```python
sns.kdeplot(tips_data,x="tip",hue="sex")
```

```python
sns.kdeplot(tips_data,x="tip",y="total_bill",hue="sex")
```
