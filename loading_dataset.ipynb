```python
import seaborn as sns
# sns.get_dataset_names()
sns.load_dataset("tips")
sns.set_theme(style="darkgrid")
```

```python
tips_data=sns.load_dataset("tips")
sns.scatterplot(tips_data,x="total_bill",y="tip",hue="sex",size="size",
                style="smoker")
```

```python
tips_data.head()
```

```python
sns.relplot(tips_data,x="total_bill",y="tip",hue="sex",size="size",
            style="smoker",col="day",col_wrap=2,kind="scatter")
# real plot can choose the kind of plot
```
